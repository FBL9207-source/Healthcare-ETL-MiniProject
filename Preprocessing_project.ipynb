#Preprocessing Pipeline

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
#split dataset
x = health_data.drop(['billing_amount'],axis=1)
y = health_data['billing_amount']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
health_data.select_dtypes(include=['object']).columns
```

```python
health_data.select_dtypes(include=['float','int']).columns
```

```python
num_cols = ['age', 'length_of_stay']
catego_cols = ['gender', 'blood_type', 'insurance_provider', 'admission_type',
       'medication', 'test_results','medical_condition']
```

```python
# combine Pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('onehot', OneHotEncoder(handle_unknown='ignore',sparse_output=False), catego_cols),
            ])
```

```python
# preprocessing pipeline
preprocessing_pipeline = Pipeline(steps=[('preprocessor', preprocessor)])
```

```python
# Fit and transform the training data
X_train_processed = preprocessing_pipeline.fit_transform(x_train, y_train)
X_test_processed = preprocessing_pipeline.transform(x_test)
```
